```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn import linear_model


df=pd.read_csv("/content/drive/MyDrive/Dataset/API_NY.GDP.PCAP.CD_DS2_en_csv_v2_121663.csv", skiprows=4)
```

```python
df.isna().sum()
```

```python
df=df.drop(columns=['Unnamed: 70','2025'])
```

```python
df.isna().sum().sum()
```

```python
df.head()
```

```python
value_vars=df.columns[4:]
df_long=df.melt(id_vars=['Country Name', 'Country Code'], value_vars=value_vars, var_name="Years",value_name='GDP')
df_long
```

```python
df_long=df_long.dropna(subset=['GDP'])
df_long
```

```python
df_long['Country Name'].unique()
```

```python
encoder = LabelEncoder()

df_long['Country_Label'] = encoder.fit_transform(
    df_long['Country Name']
)
df_long
```

```python
inputs = df_long[['Country_Label', 'Years']]
target = df_long['GDP']
```

```python
inputs.head()
```

```python
target.head()
```

```python
inputs['Years'] = inputs['Years'].astype(int)
```

```python
from sklearn import tree
model = tree.DecisionTreeRegressor()
```

```python
model.fit(inputs, target)
```

```python
model.score(inputs, target)
```

```python
model.predict([[0, 2020]])
```
